```python
import cv2
from ultralytics import YOLO
```

```python
for i in range(10):
    test_cam = cv2.VideoCapture(i)
    if test_cam.isOpened():
        print(f"Camera index {i} is available.")
        test_cam.release()
```

```python
# Load the YOLOv8 model (pretrained on COCO dataset) 
# 'yolov8n.pt'  //modeloDetectorAmenazas17/weights/best.pt //modeloSoloHandgun2/weights/best.pt //modeloDatabase23/weights/best.pt
# //modeloDataset/weights/best.pt //modeloDatasetArreglado/weights/best.pt //modeloDatasetConMenosPersonas/weights/best.pt
# //modeloDatasetConMenosPersonas200epochs/weights/best.pt
model = YOLO('modeloDatasetConMenosPersonas200epochs/weights/best.pt')  

# Initialize video capture (0 = default webcam, 1= elgato camera hub or replace with a video file path)
cap = cv2.VideoCapture(1)
print(cv2.VideoCapture(1).isOpened())
if not cap.isOpened():
    print("Error: Could not open webcam.")
    exit()
```

```python
class_colors = {
    0: (0, 0, 255),   # Red for Handgun
    1: (0, 255, 0),   # Green for Person
    2: (0, 255, 255), # Yellow for Rifle
}

while True:
    # Read a frame from the video feed
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read frame.")
        break

    # Convert the frame to RGB (YOLO expects RGB images)
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Run YOLOv8 inference on the frame
    results = model(rgb_frame)

    # Extract detected objects from results
    detections = results[0].boxes.data  # Bounding boxes and labels
    for detection in detections:
        x1, y1, x2, y2, confidence, class_id = detection
        label = model.names[int(class_id)]  # Get class name

        color = class_colors.get(int(class_id), (255, 255, 255))
        
        # Draw the bounding box
        cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), color, 2)
        # Add label and confidence
        text = f"{label} {confidence:.2f}"
        cv2.putText(frame, text, (int(x1), int(y1) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # Display the annotated frame
    cv2.imshow("YOLOv8 Real-Time Detection", frame)

    # Break the loop if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```

```python
# Release the video capture and close OpenCV windows
cap.release()
cv2.destroyAllWindows()
```
